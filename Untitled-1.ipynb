```python
import pandas as pd
import numpy as np
import os

# --- CONFIGURATION ---
input_file = 'IEEE2Bus_voltageBus1.xlsx'  # Change to 'IEEE2Bus_voltageBus1.xlsx' if needed
output_file = 'processed_data.csv'
resolution = 0.05
# ---------------------

def extract_data(file_path):
    # 1. Load the file correctly based on extension
    ext = os.path.splitext(file_path)[-1].lower()
    try:
        if ext == '.csv':
            df = pd.read_csv(file_path, encoding='latin1')
        elif ext in ['.xlsx', '.xls']:
            df = pd.read_excel(file_path)
        else:
            print(f"Unsupported file extension: {ext}")
            return
    except Exception as e:
        print(f"Error reading file: {e}")
        return

    # 2. Process and Clean
    # The first column is "time, va1" as a single string. Split it.
    # Note: Using .iloc[:, 0] to be safe with column names
    temp_split = df.iloc[:, 0].astype(str).str.split(',', expand=True)
    df['time'] = pd.to_numeric(temp_split[0], errors='coerce')
    df['va1'] = pd.to_numeric(temp_split[1], errors='coerce')
    
    # Rename the second column to v_ref
    df['v_ref'] = df.iloc[:, 1]
    
    # Drop any NaNs created during conversion
    df = df.dropna(subset=['time', 'va1'])

    # 3. Downsample to specified resolution (e.g., 0.05s)
    time_max = df['time'].max()
    target_times = np.arange(0, time_max + resolution, resolution)
    
    downsampled_list = []
    for t in target_times:
        # Find the row with the time closest to the target 't'
        idx = (df['time'] - t).abs().idxmin()
        downsampled_list.append(df.iloc[idx])
    
    final_df = pd.DataFrame(downsampled_list).drop_duplicates()

    # 4. Save for Overleaf
    final_df[['time', 'va1', 'v_ref']].to_csv(output_file, index=False)
    print(f"Successfully created {output_file} with {len(final_df)} points.")

# Run the function
extract_data(input_file)
```
